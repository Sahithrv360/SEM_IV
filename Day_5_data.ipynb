```python
import pandas as pd
import numpy as np
data = pd.read_csv('student_dropout_dataset_v3.csv')
```

```python
data['Family_Income'] = data['Family_Income'].fillna(
    data.groupby(['Parental_Education','Scholarship','Part_Time_Job'])
    ['Family_Income'].transform('median')
)

data['Family_Income'] = data['Family_Income'].fillna(
    data['Family_Income'].median()
)

data['Study_Hours_per_Day'] = data['Study_Hours_per_Day'].fillna(
    data.groupby(['Semester','Scholarship','Part_Time_Job','Department','Gender'])
    ['Study_Hours_per_Day'].transform('median')
)

data['Parental_Education'] = data['Parental_Education'].fillna(
    data['Parental_Education'].mode()[0]
)

data['Stress_Index'] = data['Stress_Index'].fillna(
    data['Stress_Index'].median()
)

data['Semester'] = data['Semester'].str.extract('(1|2|3|4)').astype(int)
```

```python

gender = data['Gender'].unique().tolist()
gender_mapping = {g : i for i,g in enumerate(gender)}
print(gender_mapping)

education = data['Parental_Education'].unique().tolist()
education_mapping = {edu : i for i,edu in enumerate(education)}
print(education_mapping)

sch = data['Scholarship'].unique().tolist()
sch_mapping = {s : i for i,s in enumerate(sch)}
print(sch_mapping)

job = data['Part_Time_Job'].unique().tolist()
job_mapping = {j: i for i,j in enumerate(job)}
print(job_mapping)

dept = data['Department'].unique().tolist()
dept_mapping = {d : i for i,d in enumerate(dept)}
print(dept_mapping)
```

```python
# Inverse Mapping Creation File

# Gender Inverse Mapping
gender_invmap = {i: g for g, i in gender_mapping.items()}

# Parental Education Inverse Mapping
education_invmap = {i: edu for edu, i in education_mapping.items()}

# Scholarship Inverse Mapping
sch_invmap = {i: s for s, i in sch_mapping.items()}

# Part-Time Job Inverse Mapping
job_invmap = {i: j for j, i in job_mapping.items()}

# Department Inverse Mapping
dept_invmap = {i: d for d, i in dept_mapping.items()}

print("Inverse Mappings Created Successfully")
```

```python
# Apply Mappings to DataFrame

# Apply Gender Mapping
data['Gender'] = data['Gender'].map(gender_mapping)

# Apply Parental Education Mapping
data['Parental_Education_label'] = data['Parental_Education'].map(education_mapping)

# Apply Scholarship Mapping
data['Scholarship'] = data['Scholarship'].map(sch_mapping)

# Apply Part-Time Job Mapping
data['Part_Time_Job'] = data['Part_Time_Job'].map(job_mapping)

# Apply Department Mapping
data['Department_label'] = data['Department'].map(dept_mapping)

data['Age'] = data['Age'].astype(int)
print("All mappings applied successfully")
```

```python
data.drop(columns=['Student_ID','Semester_GPA','GPA','Internet_Access',
                   'Assignment_Delay_Days',
                  ],inplace=True)
```
