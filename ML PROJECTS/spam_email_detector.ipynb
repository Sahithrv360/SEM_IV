```python
from sklearn.feature_extraction.text import CountVectorizer # to convert values to numeric data / vectors

from sklearn.model_selection import train_test_split # to split the dataset 

from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import accuracy_score

import numpy as np

import pandas as pd
```

```python
file_path = "C:/Users/sahit/OneDrive/Desktop/coding/IML/mail_data.csv"
data = pd.read_csv(file_path)
print(data.head())
```

```python
print(data.tail())
```

```python
print(data.isnull().sum())
```

```python
vectorizer = CountVectorizer()
hello = vectorizer.fit_transform(data['Message'])
```

```python
X_train,X_test,Y_train,Y_test = train_test_split(hello,data['Category'],test_size=0.2)
```

```python
model = MultinomialNB()
model.fit(X_train,Y_train)
```

```python
yPred = model.predict(X_test)

#accuracy measure

accuracy = accuracy_score(Y_test,yPred)
```

```python
print(accuracy)
```

```python
def Spam_Prediction(message):
    messageVectorizer = vectorizer.transform([message])
    prediction = model.predict(messageVectorizer)
    return 'Spam' if prediction[0] == 1 else 'Ham'
```

```python
msg = input('Enter a email message to predict it whether it is a spam or not :')
predicted = Spam_Prediction(msg)
print(predicted)
```
