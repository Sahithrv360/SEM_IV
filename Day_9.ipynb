```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score, davies_bouldin_score
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv('ultimate_student_productivity_dataset_5000.csv')
X = data.select_dtypes(include=np.number)
X = X.drop(columns=['student_id', 'exam_score'])
```

```python
def performance(obj, msg):
    labels = obj.fit_predict(X)
    print(f"\n{msg}")
    print("Silhouette Score:", silhouette_score(X, labels))
    print("Davies-Bouldin Index:", davies_bouldin_score(X, labels)) 
```

```python
print("🔹 K-Means Tuning")
for k in range(2, 5):
    kmeans = KMeans(n_clusters=k, init='k-means++', random_state=42)
    performance(kmeans, f"K-Means (k={k})")
```

```python
print("\n🔹 Agglomerative Tuning")
for linkage in ['ward', 'complete', 'average']:
    agg = AgglomerativeClustering(n_clusters=3, linkage=linkage)
    performance(agg, f"Agglomerative (linkage={linkage})")
```
