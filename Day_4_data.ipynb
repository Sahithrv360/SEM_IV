```python
import pandas as pd

data = pd.read_csv('ultimate_student_productivity_dataset_5000.csv')
```

```python
data['total_study_hours'] = data['study_hours']+data['self_study_hours']+data['online_classes_hours']
data['total_smart_device_usage_hrs'] = data['online_classes_hours']+data['social_media_hours']+data['gaming_hours']
data['avg_screen_time'] = (data['total_smart_device_usage_hrs']+data['screen_time_hours'])/2
```

```python
gender = data['gender'].unique().tolist()
gender_mapping = {g : i for i,g in enumerate(gender)}
print(gender_mapping)

education = data['academic_level'].unique().tolist()
education_mapping = {edu : i for i,edu in enumerate(education)}
print(education_mapping)

internet = data['internet_quality'].unique().tolist()
internet_mapping = {inte : i for i,inte in enumerate(internet)}
print(internet_mapping)
```

```python
inv_gender_mapping = {v : k for k,v in gender_mapping.items()}
print(inv_gender_mapping)

inv_education_mapping = {v : k for k,v in education_mapping.items()}
print(inv_education_mapping)

inv_internet_mapping = {v : k for k,v in internet_mapping.items()}
print(inv_internet_mapping)
```

```python
data['Gender_label'] = data['gender'].map(gender_mapping)

data['education_label'] = data['academic_level'].map(education_mapping)

data['internet_label'] = data['internet_quality'].map(internet_mapping)
```

```python
data.columns
```

```python
import numpy as np
from scipy.stats import zscore
# Calculate Z-score
X_mis = data
z_scores = np.abs(zscore(data.columns))
# Threshold
threshold = 3

# Detect outliers
outliers = data[(z_scores<threshold).all(axis=1)]

print("Total outliers:", outliers.shape)
outliers.head()
```

```python
data.to_csv('student_chart.csv',index=False)
```
