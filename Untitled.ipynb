```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score

# Create dataset
X, y = make_blobs(n_samples=300, centers=3, random_state=42)

# 1. K-Means
kmeans = KMeans(n_clusters=3, random_state=42)
k_labels = kmeans.fit_predict(X)

# 2. Hierarchical (Agglomerative)
agg = AgglomerativeClustering(n_clusters=3)
a_labels = agg.fit_predict(X)

# 3. Divisive (using KMeans as approximation)
div = KMeans(n_clusters=3, random_state=1)
d_labels = div.fit_predict(X)

# Evaluation
print("K-Means Silhouette Score:", silhouette_score(X, k_labels))
print("Agglomerative Score:", silhouette_score(X, a_labels))
print("Divisive Score:", silhouette_score(X, d_labels))

# Plot
plt.figure(figsize=(12,4))

plt.subplot(131)
plt.scatter(X[:,0], X[:,1], c=k_labels)
plt.title("K-Means")

plt.subplot(132)
plt.scatter(X[:,0], X[:,1], c=a_labels)
plt.title("Agglomerative")

plt.subplot(133)
plt.scatter(X[:,0], X[:,1], c=d_labels)
plt.title("Divisive")

plt.show()
```
